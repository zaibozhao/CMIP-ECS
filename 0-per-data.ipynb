# CMIP5

## bcc-csm1-1

```python
data_address = '../ECS/data/raw/tas_Amon_bcc-csm1-1_historical_r1i1p1_185001-201212.nc'
ncData = nc.Dataset(data_address,'r',format = 'NETCDF4')
tas = ncData.variables['tas'][:]
lat = ncData.variables['lat'][:]
    
tas_year = tas.reshape(-1,12,64,128).mean(axis=1)
w = np.cos(np.deg2rad(lat))

global_year = (
    (tas_year * w[None,:,None]).sum(axis=(1,2))
    / (w.sum() * tas_year.shape[2])
)

baseline = global_year[11:50].mean()

baseline_tas = global_year - baseline
final_tas = baseline_tas[50:156]

save_address = 'data/CMIP5_annual_tas.h5py'
f_w = h5py.File(save_address,'w')
f_w['bcc-csm1-1'] = final_tas
f_w.close()
```

## GFDL-CM3

```python
root_address = '../ECS/data/raw/GFDL-CM3'
paths = [os.path.join(root_address, x) for x in os.listdir(root_address) if x.endswith('.nc')]
for i in range (len(paths)):
    if  i == 0:
        data_address = paths[i]
        ncData = nc.Dataset(data_address,'r',format = 'NETCDF4')
        tas = ncData.variables['tas'][:]
    else:
        data_address = paths[i]
        ncData = nc.Dataset(data_address,'r',format = 'NETCDF4')
        tas = np.concatenate((tas, ncData.variables['tas'][:]), axis=0)
    
lat = ncData.variables['lat'][:]
tas_year = tas.reshape(-1,12,tas.shape[1],tas.shape[2]).mean(axis=1)
w = np.cos(np.deg2rad(lat))
global_year = (
    (tas_year * w[None,:,None]).sum(axis=(1,2))
    / (w.sum() * tas_year.shape[2])
)
baseline = global_year[1:40].mean()
baseline_tas = global_year - baseline
final_tas = baseline_tas[40:]

save_address = 'data/CMIP5_annual_tas.h5py'
f_w = h5py.File(save_address,'a')
f_w['GFDL-CM3'] = final_tas
f_w.close()
```

# CMIP6

```python
folder = '../ECS/data/raw/cmip6'

model_paths = [
    os.path.join(folder, name)
    for name in os.listdir(folder)
    if os.path.isdir(os.path.join(folder, name))
]

model_name_dirs = [
    d for d in os.listdir(folder)
    if os.path.isdir(os.path.join(folder, d))
]

for model_num in range(len(model_paths)):

    model_path = model_paths[model_num]
    model_name = model_name_dirs[model_num]
    data_paths = [os.path.join(model_path, x) for x in os.listdir(model_path) if x.endswith('.nc')]
    for i in range (len(data_paths)):
        if  i == 0:
            data_address = data_paths[i]
            ncData = nc.Dataset(data_address,'r',format = 'NETCDF4')
            tas = ncData.variables['tas'][:]
        else:
            data_address = data_paths[i]
            ncData = nc.Dataset(data_address,'r',format = 'NETCDF4')
            tas = np.concatenate((tas, ncData.variables['tas'][:]), axis=0)

    lat = ncData.variables['lat'][:]
    tas_year = tas.reshape(-1,12,tas.shape[1],tas.shape[2]).mean(axis=1)
    w = np.cos(np.deg2rad(lat))
    global_year = (
        (tas_year * w[None,:,None]).sum(axis=(1,2))
        / (w.sum() * tas_year.shape[2])
    )
    baseline = global_year[11:50].mean()
    baseline_tas = global_year - baseline
    final_tas = baseline_tas[50:]
    save_address = 'data/CMIP6_annual_tas.h5py'
    f_w = h5py.File(save_address,'a')
    f_w[model_name] = final_tas
    f_w.close()

# CESM2
```
